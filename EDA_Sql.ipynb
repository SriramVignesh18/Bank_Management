```python
vendor_sales_summ = pd.read_sql_query("""
WITH FreightSummary AS (
    SELECT
        VendorNumber,
        SUM(Freight) AS FreightCost
    FROM vendor_invoice
    GROUP BY VendorNumber
),
PurchaseSummary AS (
    SELECT
        p.VendorNumber,
        p.VendorName,
        p.Brand,
        p.Description,
        p.PurchasePrice,
        pp.Price AS ActualPrice,
        pp.Volume,
        SUM(p.Quantity) AS TotalPurchaseQuantity,
        SUM(p.Dollars) AS TotalPurchaseDollars
    FROM purchases AS p
    JOIN purchase_prices AS pp ON p.Brand = pp.Brand
    WHERE p.PurchasePrice > 0
    GROUP BY
        p.VendorNumber,
        p.VendorName,
        p.Brand,
        p.Description,
        p.PurchasePrice,
        pp.Price,
        pp.Volume
),
SalesSummary AS (
    SELECT
        s.VendorNo,
        s.Brand,
        SUM(s.SalesQuantity) AS TotalSalesQuantity,
        SUM(s.SalesDollars) AS TotalSalesDollars,
        SUM(s.SalesPrice) AS TotalSalesPrice,
        SUM(s.ExciseTax) AS TotalExciseTax
    FROM sales AS s
    GROUP BY s.VendorNo, s.Brand
)
SELECT
    ps.VendorNumber,
    ps.VendorName,
    ps.Brand,
    ps.Description,
    ps.PurchasePrice,
    ps.ActualPrice,
    ps.Volume,
    ps.TotalPurchaseQuantity,
    ps.TotalPurchaseDollars,
    ss.TotalSalesQuantity,
    ss.TotalSalesDollars,
    ss.TotalSalesPrice,
    ss.TotalExciseTax,
    fs.FreightCost
FROM PurchaseSummary AS ps
LEFT JOIN SalesSummary AS ss
    ON ps.VendorNumber = ss.VendorNo
    AND ps.Brand = ss.Brand
LEFT JOIN FreightSummary AS fs
    ON ps.VendorNumber = fs.VendorNumber
ORDER BY ps.TotalPurchaseDollars DESC
""", con)
```

```python
vendor_sales_summ.isnull().sum()
```

```python
vendor_sales_summ['Volume']=vendor_sales_summ['Volume'].astype('float64')
```

```python
vendor_sales_summ['VendorName']=vendor_sales_summ['VendorName'].str.strip()
```

```python
vendor_sales_summ.fillna(0,inplace=True)
```

```python
vendor_sales_summ['Profit']=vendor_sales_summ['TotalSalesDollars']-vendor_sales_summ['TotalPurchaseDollars']
```

```python
vendor_sales_summ['ProfitMargin']=(vendor_sales_summ['Profit']/vendor_sales_summ['TotalSalesDollars'])*100
```

```python
vendor_sales_summ['StockTurnover']=(vendor_sales_summ['TotalSalesQuantity']/vendor_sales_summ['TotalPurchaseQuantity'])
```

```python
vendor_sales_summ['S2PRatio']=(vendor_sales_summ['TotalSalesDollars']/vendor_sales_summ['TotalPurchaseDollars'])
```

```python
vendor_sales_summ.columns
```

```python
cursor=con.cursor()
cursor.execute("""CREATE TABLE IF NOT EXISTS Vendor_Sales_Summary(
VendorNumber INT, VendorName VARCHAR(150), Brand INT, Description VARCHAR(150), PurchasePrice DECIMAL(10,2),
ActualPrice DECIMAL(10,2), Volume, TotalPurchaseQuantity INT, TotalPurchaseDollars DECIMAL(12,2),
TotalSalesQuantity DECIMAL(12,2), TotalSalesDollars DECIMAL(12,2), TotalSalesPrice DECIMAL(12,2), 
TotalExciseTax DECIMAL(12,2), FreightCost DECIMAL(12,2), Profit DECIMAL(12,2), ProfitMargin DECIMAL(12,2),
StockTurnover DECIMAL(12,2), S2PRatio DECIMAL(12,2), Primary Key (VendorNumber,Brand)
);""")
```

```python
vendor_sales_summ.to_sql("Vendor_Sales_Summary",con,if_exists='replace',index=False)
pd.read_sql_query("select * from Vendor_Sales_Summary",con)
```
